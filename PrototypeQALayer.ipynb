```python
# 6. Prototype Question Parser for Graph QA
import spacy
nlp_qa = spacy.load("en_core_web_sm")

sample_q = "Which chemicals cause alopecia?"
doc = nlp_qa(sample_q)

wh_word = [tok.text for tok in doc if tok.tag_ in ("WP", "WRB")]
verbs = [tok.lemma_ for tok in doc if tok.pos_ == "VERB"]
print("Question Type:", wh_word, "| Verbs:", verbs)

# Use SciSpaCy NER to extract biomedical entities
import scispacy
ner = spacy.load("en_ner_bc5cdr_md")
ents = ner(sample_q).ents
print("Entities:", [(e.text, e.label_) for e in ents])

# Output:
# Entities: [('alopecia', 'DISEASE')]
# => Translate to Cypher: MATCH (c:Chemical)-[:ASSOCIATED_WITH]->(d:Disease {name:"Alopecia"}) RETURN c.name
```
